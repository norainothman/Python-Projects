```python
# Loops
# Loops in Python are used to repeat a block of code multiple times
# without writing the same code again and again.
# Loops help us repeat tasks efficiently and smartly.
# A loops is a control structure that repeats a block of code until a condition is met.
# Loops got 2 category:
# for/while

# for Loops
# it is used to iterate over a sequence [string,list,tuple,array,etc...]

# for iterable_variable in Sequence :
#  code to be executed

# AFTER EVERY ITERATION BODY OF FOR LOOP WILL BE EXECUTED

for i in 'Python':
  print(i)
```

```python
# range() function
# Used to generate sequence of integer values

# range(x)   --> must be + int number
# starting from Zero till x-1

print(list(range(11)))

# range(x,y)
# x number must be smaller that y number
# it is used to generate sequence of integer values
# starting from x till y-1

print(list(range(1,11)))

# range(x,y,stepCount)
# x number must be smaller that y number
# it is used to generate sequence of integer values
# starting from x till y-1 based on stepCount

print(list(range(1,11,3)))

# using range function print all even numbers till 100

print(list(range(0,101,2)))
```

```python
# NESTED FOR LOOP
# for loop defined inside another for loop

colors = ['red','blue','green']
vehicles = ['car','bus','bike']

for c in colors:
  for v in vehicles:
    print(c,v)
```

```python
# Star Pattern (2D Array)
# Using nested for loop

rows = 5
for i in range (1,rows+1): # i in range (1,6)  i --> 1 2 3 4 5
  for j in range(i):       # j in range (0)
    print('*',end=' ')
  print()
```

```python
# While loop
# While loop repeats a block of code until certain condition met

# syntax
# initialization
# while condition:
#  a block of code
#  increment/decrement

# While checks the condition first
# if the condition is TRUE hile will execute the block of code
# after executing the block of code while will check the condition again
i = 1
while i<=3: # 1<=3 (t)   2<=3(t)   3<=3(t)   4<=3(f)
  print(i)  # 1          2         3
  i+= 1     # i=1+1(2)   i=2+1(3)  i=3+1(4)
```

```python
# print mathemical table of a number using while loop

num = int(input('Enter Any Number : '))
i=1
while i<=10:
  print(f'{num} x {i} = {num*i}')
  i+=1
```

```python
# break and continue statement
# break statement
# it is used to terminate the execution of a LOOP

# break/continue should be defined inside decision making

for i in range (1,11):
  if  i == 5:
    break
  print(i)
```

```python
# continue statement
# it is used to skip the current execution of a LOOP

for i in range (1,11) :
  if i == 5 or i == 9 :
    continue
  print(i)
```
