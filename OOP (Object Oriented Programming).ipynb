```python
# OOPS
# OOP stands for Object-Oriented Programming.
# Object-oriented programming (OOP) is a programming paradigm based on the
# concept of "objects"
 
# Procedural programming is about writing procedures or methods that perform operations
# on the data, while object-oriented programming is about creating objects
# that contain both data and methods.
 
# Object-oriented programming has several advantages over procedural programming:
 
# a.) OOP is faster and easier to execute
# b.) OOP provides a clear structure for the programs
# c.) OOP helps to keep the Java/Python code DRY "Don't Repeat Yourself", and makes the code easier
#     to maintain, modify and debug
# d.) OOP makes it possible to create full reusable applications with less code and
#     shorter development time
 
# Class
# A class is a blueprint that defines the variables and the methods common to all objects.
# a class is a collection of methods and variables
 
 
# Object
# An object is called an instance of a class.
 
 
# Self Keyword
# The self parameter is a reference to the current instance of the class, and is used to
# access variables that belongs to the class.
# The first parameter of each method in a class must be the self ,
# which refers to the calling object. However, you can give any name to the first
# parameter, not necessarily self.
 
# __init__ in Python
#  Constructors are used to initializing the object’s state. The task of constructors is
# to initialize(assign values) to the data members of the class when
# an object of the class is created.
# The __init__() function is called automatically every time the class is being used to create
# a new object.
```

```python
# Class in Python 
# Class is a collection of methods/functions and variables
# In Python a class can be defined by using a class keyword.

class myClass:

    '''
    This is my first class in python
    I have just started learning OOPs
    '''
    a = 10          # member variable
    b = 20          # member variable

    def func1():                                  # member function
        print(' I am the method of a class')    
    def func2():                                  # member function
        print('I am another method of a class')
```

```python
class myClass:
    a = 10          
    b = 20        

    def func1():                                
        print(' I am the method of a class')    
    def func2():                                 
        print('I am another method of a class')

# RECOMMENDED APPROACH
# In OOPs in order to access class variables and functions
# We need to define the Object of a class

# Syntax to define object
# objName = className()

# Object can access class variables and functions
# class can have n number of objects

obj1 = myClass()  # Creating object of a class
obj2 = myClass()  # Creating another object of class

print(obj1.a)
print(obj1.b)

print(obj2.a)
print(obj2.b)
```

```python
class myClass:  # 3) giving your class clear idea that now obj1 want to use this function
    a = 10          
    b = 20        

    def func1(self):  # 2) self value will be replace by current calling object  func1(obj1)                     
        print(' I am the method of a class')    
    def func2(self):                                 
        print('I am another method of a class')
        
obj1 = myClass() 
obj2 = myClass() 
obj3 = myClass() 

obj1.func1()  # 1) here i am calling this func1 with obj1
obj2.func1()
obj3.func1()
```

```python
# Task
# We need to take 3 information from 3 users
# user1/user2/user3
# name = ?
# email = ?
# fon no = ?

# Task 1 : take 3 information from 3 users
# Task 2 : print these information
# Task 3 : print only information of user2 
# Task 4 : print email and fon no from user3

# USING PROCEDURAL PROGRAMMING
# Task 1 : take 3 information from 3 users

users = []
for i in range (1,4):
    print(f'Enter Details For User {i}')  
    name = input('Enter Your Name : ')
    email = input('Enter Your Email : ')
    fonno = input('Enter Your Fon No : ')
    
    # storing information 
    user_data = [name,email,fonno]
    users.append(user_data)

# Task 2 : print these information
print(users)

# Task 3 : print only information of user2 
print(users[1])

# Task 4 : print email and fon no from user3
print(users[2][1])
print(users[2][2])
```

```python
class userDetails:

# Use __init__() when u want to assign values to a class variables
# this __init__() call itself, the moment we create the object of a class

    def __init__(self,n,e,f):
        self.name = n
        self.email = e
        self.fonno = f

    def printUserDetails(self):
        print(self.name,self.email,self.fonno)

# While creating an object now we need to create arguments

us1 = userDetails('John','j@gmail.com',6089563651)
us2 = userDetails('Alex','al@gmail.com',608974369)
us3 = userDetails('Smith','sm@gmail.com',606695879)

print('\n========Printing All Information========\n')
us1.printUserDetails()
us2.printUserDetails()
us3.printUserDetails()

print('\n========Printing 2nd user Information========\n')
us2.printUserDetails()

print('\n========Printing 3rd user Email & fonno ========\n')
print(us3.email)
print(us3.fonno)
```
