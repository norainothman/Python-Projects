```python
# if
# if used to check the condition first
# if the condition is TRUE it will execute the block of code

# Syntax
# if condition:
#   block of code

age = int(input("Enter your age: "))
if age >= 18:
  print("You are eligible to vote")
```

```python
num = int(input("Enter a number: "))
if num % 2 == 0:
  print('Checking The Number')
  print("The number is even")
  print('Done')
  print('Still I am in IF Block')
print('Rest of the code')
```

```python
a = 19
b = 45
c = 21
d = 141
e = 67

largest = 0
if a>largest:     # 19>0 (t)
  largest = a     # largest = 19
if b>largest:     # 45>19 (t)
  largest = b     # largest = 45
if c>largest:     # 21>45 (f)
  largest = c     # largest = 45
if d>largest:     # 141>45 (t)
  largest = d     # largest = 141
if e>largest:     # 67>141 (f)
  largest = e     # largest = 141
print(f'Largest Nummber is : ',largest)
```

```python
# if else
# if used to check the condition first
# if the condition is TRUE it will execute the block of code
# if the condition is FALSE, then else part will be executed

age = int(input('Enter Your Age : '))
if age >= 18:
  print('Welcome User')
  print('You are eligible to Play The Game')
  print('Game Is Loading.......')
else:
  print('Sorry User')
  print('You are not eligible to Play The Game')
  print('Game Is Not Loading')
```

```python
# if elif else
# ONLY ONE BLOCK OF CODE WILL BE EXECUTED AT A TIME

num = int(input('Enter Any Number : '))
if num > 0:
  print('Number is Positive')
elif num == 0:
  print('Number is Zero')
else:
  print('Number is Negative')
```

```python
marks=int(input('Enter Your Marks : '))
if marks <=100 and marks >=85 :
  print('Grade A')
elif marks <=84 and marks >=70 :
  print('Grade B')
elif marks <=69 and marks >=55 :
  print('Grade C')
elif marks <=54 and marks >=40 :
  print('Grade D')
elif marks <=39 and marks >=0 :
  print('Grade F')
else:
  print('Invalid Input')
```
