```python
# Guess a number game
# ypur machine will generate a random number
# we need to guess that number in 4 attempts

import random
guessAttempt = 1
print('Guess A Number Game 🤖 ')
print('Enter Your Name')
name = input()
print(f'Welcome User {name} 👋🏼 ')
print('I am generating one random number between 1 till 13. Please wait.... ')
num = random.randint(1,13)
print('Number Is Generated ❗ ')
print('Guess The Number in 4 Attempts ')

while guessAttempt <=4 :
  print(f'\n{guessAttempt} Attempt. Enter Your Guest')
  guess = int(input())

  if guess == num :
    print(f'Congratulation 🎊 You Guess The Number Right In {guessAttempt} Attempts.')
    break
  if guess > num :
    print('Your Guess Is Too High ⬆️ Think of a Lower Number.\n')
  if guess < num :
    print('Your Guess Is Too Low ⬇️ Think of a Higher Number.\n')

  guessAttempt = guessAttempt + 1

if guess != num :
  print ('No More Attempts left. ')
  print ('GAME OVER')
  print ('The Number Was 🎲 ',num)
  print ('Better Luck Next Time')

```
