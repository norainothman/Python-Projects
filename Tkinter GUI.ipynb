```python
# Guess a number game

from tkinter import *
import random
w = Tk()
w.title('Guess A Number Widget')
w.geometry('600x400')

# background color for window
w.config(bg='#065569')
w.resizable(width=False,height=False)

# Main Logic
ranNum = random.randint(1,10)
chance = 4
displayResult = StringVar()

def Check_guess():
    global ranNum
    global chance
    userInput = int(user_input.get())
    if chance > 0 :
        if userInput == ranNum:
            msg = 'Congratulation! You Won ' + str(ranNum)+ ' is the right number'
            user_input.config(state='disabled')
            guess_button.config(state='disabled')
        elif userInput>ranNum:
            chance = chance - 1
            msg = 'Your Guess Is TOO HIGT. Think of a Smaller Number'
            user_input.delete(0,END)
        elif userInput<ranNum:
            chance = chance - 1 
            msg = 'Your Guess Is TOO LOW. Think of a Higher Number'
            user_input.delete(0,END)
    else:
        msg = 'Game Over! '+str(chance)+ ' Attempts Left. The Correct Number was '+str(ranNum)
        user_input.config(state='disabled')
        guess_button.config(state='disabled')
    displayResult.set(msg)

# Create Widgets
title = Label(w,text='Guess A Number Game',font=('Arial',20),fg='#fffcbd',bg='#065569')
gameInstruction = Label(w,text='Guess The Number Between 1 to 10 in 4 Attempts',
                        font=('Arial',12),fg='#fffcbd',bg='#065569')
user_input = Entry(w,font=('Arial',20))
guess_button = Button(w,text='Guess',font=('Arial',13),fg='#13d675',bg='black',command=Check_guess)
exit_button = Button(w,text='Exit Game',font=('Arial',14),fg='white',bg='#b82741',command=w.destroy)
outputLabel = Label(w,font=('Arial',14),fg='#fffcbd',bg='#065569',textvariable=displayResult)

# Place Widgets
title.place(x=140,y=50)
gameInstruction.place(x=95,y=95)
user_input.place(x=180,y=150)
guess_button.place(x=420,y=150)
exit_button.place(x=300,y=300)
outputLabel.place(x=100,y=220)

w.mainloop()
```

```python
# What is Sentiment Analysis?

# Process of analyzing digital text.
# To determine if the emotional tone of the message is +ve, -ve or neutral. 
# large volumes of text data (emails, cust support chat transcripts, social media comments and reviews.)
# It can scan text automatically to determine cust attitude towards a topic.
# This can improve cust service and increase brand reputation.

# VADER (Valence Aware Dictionary and Sentiment Reasoner.)
# Tool used for sentiment analysis. 
# Figure out if text is expressing +ve, -ve or neutral emotions.
# It assigns a score between -1 (most -ve) and +1 (most +ve) to show how strong the sentiment is.
# Research suggest VADDER achieves score 0.96 for sentiment classification.

# SentimentIntensityAnalyzer class provides a method called polarity_score()
# It takes a piece of text as input and returns a dictionary containing sentiment scores for the text.
# Dictionary contains four keys : neg, neu, pos and compound. 
```

```python
# Sentiment Analyzer App

from textblob import TextBlob
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
sid_obj = SentimentIntensityAnalyzer()

w = Tk()
w.title('Sentiment Analyzer')
w.geometry('600x400')

# background color for window
w.config(bg='#6482AD')
w.resizable(width=False,height=False)

# Main Logic
displayResult = StringVar()
def check_sentiment():
    userInput = user_input.get()
    Result = TextBlob(userInput)
    Result = sid_obj.polarity_scores(Result)

    if Result['compound'] >= 0.5:
        displayResult.set('Positive Text')
        user_input.delete(0,END)
    elif Result['compound']<=0.5:
        displayResult.set('Negative Text')
        user_input.delete(0,END)
    else:
        displayResult.set('Neutral Text')
        user_input.delete(0,END)
        
# Create Widgets
title = Label(w,text='Sentiment Analyzer. Please Insert Some Text Below',font=('Arial',14),
              fg='#F5EDED',bg='#6482AD')
user_input = Entry(w,font=('Arial',14))
button1 = Button(w,text='Click Here To Analyze', font=('Arial',14),fg='#7C00FE',bg='#BC9F88',
                command=check_sentiment)
outputLabel = Label(w,font=('Arial',18),fg='#522258',bg='#6482AD',textvariable=displayResult)
exit_button = Button(w,text='Exit',font=('Arial',14),fg='white',bg='#b82741',command=w.destroy)

# Place Widgets
title.place(x=60,y=50)
user_input.place(x=150,y=120)
button1.place(x=390,y=120)
outputLabel.place(x=250,y=200)
exit_button.place(x=300,y=300)

w.mainloop()
```
